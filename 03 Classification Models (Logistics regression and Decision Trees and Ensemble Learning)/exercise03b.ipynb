This is the same file as tutorial 03b. The first 75% of the rows train the model. The last 25% are the test set. Numeric columns stay in their original units. Category columns are turned into zeros and ones.

`X_train` and `X_test` are those columns. `y_train` and `y_test` are `active`: 1 for a walk or a bike trip, and 0 otherwise.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

trips = pd.read_csv("mode_choice.csv")
n_train = int(len(trips) * 0.75)
train = trips.iloc[:n_train].copy()
test = trips.iloc[n_train:].copy()
print("train", len(train), "rows")
print("test ", len(test), "rows")

num = ["distance_km", "travel_min", "start_hour", "hh_vehicles", "hh_bikes"]
cat = ["purpose", "day_type", "age_group", "sex", "car_licence", "employed"]
X_train = pd.concat([
    train[num],
    pd.get_dummies(train[cat], dtype=int),
], axis=1)
X_test = pd.concat([
    test[num],
    pd.get_dummies(test[cat], dtype=int),
], axis=1)
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
y_train = train["active"]
y_test = test["active"]
print("columns", X_train.shape[1])
```
