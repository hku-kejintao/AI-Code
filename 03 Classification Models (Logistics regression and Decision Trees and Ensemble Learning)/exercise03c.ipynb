```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

trips = pd.read_csv("mode_choice.csv")
n = len(trips)
train = trips.iloc[: int(n * 0.50)].copy()
valid = trips.iloc[int(n * 0.50) : int(n * 0.75)].copy()
test = trips.iloc[int(n * 0.75) :].copy()
print("train", len(train), "rows")
print("valid", len(valid), "rows")
print("test ", len(test), "rows")

num = ["distance_km", "travel_min", "start_hour", "hh_vehicles", "hh_bikes"]
cat = ["purpose", "day_type", "age_group", "sex", "car_licence", "employed"]
X_train = pd.concat([
    train[num],
    pd.get_dummies(train[cat], dtype=int),
], axis=1)
X_valid = pd.concat([
    valid[num],
    pd.get_dummies(valid[cat], dtype=int),
], axis=1)
X_valid = X_valid.reindex(columns=X_train.columns, fill_value=0)
X_test = pd.concat([
    test[num],
    pd.get_dummies(test[cat], dtype=int),
], axis=1)
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
y_train = train["mode"]
y_valid = valid["mode"]
y_test = test["mode"]
print("columns", X_train.shape[1])
```
